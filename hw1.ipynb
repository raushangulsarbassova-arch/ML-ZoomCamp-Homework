```python
X = new_df.to_numpy()
X 
```

```python
XT = X.T
XT
```

```python
XTX = X.T @ X
XTX
```

```python
import numpy as np
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
w = XTX_inv @ X.T @ y
w
```
